# MineSites — SATF Business Viability Simulation
## Swarm-Augmented Analysis: Will It Work?

We apply the **Swarm-Augmented Time Series Forecasting (SATF)** framework to stress-test the MineSites business concept.

**The concept:** A private, verified professional network for the mining and metals industry. Three surfaces — People, Problems, Work. No feeds, no vanity metrics, no noise. Trust-gated access, reputation-based matching, escrow-protected engagements.

**The simulation:** 8 domain experts with conflicting perspectives debate the concept across 5 rounds, shifting positions as arguments develop. We extract sentiment trajectories, identify consensus and divergence, and produce a unified viability assessment.

**The agents:**

| Agent | Role | Natural Bias |
|---|---|---|
| Margaret Osei | 30-year mine operations veteran, Ghana/Australia | Practical skeptic — "show me it works underground" |
| James Thornton | VC partner, platform investments | Pattern-matcher — compares to every marketplace he's seen fail |
| Dr. Kenji Watanabe | Independent mining consultant, 15 years solo | The target user — would he actually pay for this? |
| Rachel Singh | Mining recruitment specialist | Knows the talent pain — sees the real demand signal |
| Tom Fitzgerald | FIFO drilling supervisor, WA | Ground-level reality check — will workers actually use this? |
| Catherine Moreau | Former AusIMM executive | Industry politics expert — knows why past attempts failed |
| Diego Vargas | Founder of a failed mining community app | Been there, burned — knows exactly what kills these products |
| Priya Mehta | Mining-focused PE analyst | Evaluates unit economics, TAM, and defensibility |

## Sentiment Trajectory Analysis

The swarm shifted from **skeptical** (Round 1 avg: 4.5/10) to **cautiously bullish** (Round 5 avg: 6.25/10). Critically, no agent moved backwards — even the strongest skeptics (Diego, Tom) shifted upward as the strategy crystallised during debate.

```python
rounds = np.arange(1, 6)

# --- Figure 1: Individual Agent Trajectories ---
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Top-left: All agent sentiment trajectories
colors = ["#d32f2f", "#1976d2", "#388e3c", "#f57c00", "#7b1fa2", "#00796b", "#c2185b", "#455a64"]
for i, agent in enumerate(AGENTS):
    axes[0, 0].plot(
        rounds,
        SENTIMENT[agent],
        "o-",
        color=colors[i],
        label=agent.split()[0],
        linewidth=2,
        markersize=6,
    )
axes[0, 0].axhline(y=5.5, color="gray", linestyle="--", alpha=0.5, label="Neutral")
axes[0, 0].set_xlabel("Debate Round")
axes[0, 0].set_ylabel("Sentiment (1-10)")
axes[0, 0].set_title("Individual Agent Sentiment Trajectories")
axes[0, 0].legend(fontsize=7, loc="lower right")
axes[0, 0].set_ylim(0, 10)
axes[0, 0].grid(True, alpha=0.3)

# Top-right: Swarm average + confidence-weighted average
avg_sentiment = [np.mean([SENTIMENT[a][r] for a in AGENTS]) for r in range(5)]
weights_per_round = [np.array([CONFIDENCE[a][r] / 100.0 for a in AGENTS]) for r in range(5)]
weighted_sentiment = [
    np.average([SENTIMENT[a][r] for a in AGENTS], weights=weights_per_round[r]) for r in range(5)
]
axes[0, 1].plot(rounds, avg_sentiment, "ro-", linewidth=2.5, markersize=8, label="Average")
axes[0, 1].plot(
    rounds, weighted_sentiment, "bs-", linewidth=2.5, markersize=8, label="Confidence-Weighted"
)
axes[0, 1].axhline(y=5.5, color="gray", linestyle="--", alpha=0.5)
axes[0, 1].fill_between(
    rounds, 5.5, avg_sentiment, alpha=0.15, color="green", where=[s > 5.5 for s in avg_sentiment]
)
axes[0, 1].fill_between(
    rounds, avg_sentiment, 5.5, alpha=0.15, color="red", where=[s < 5.5 for s in avg_sentiment]
)
axes[0, 1].set_xlabel("Debate Round")
axes[0, 1].set_ylabel("Sentiment (1-10)")
axes[0, 1].set_title("Swarm Consensus: From Skeptical to Cautiously Bullish")
axes[0, 1].legend()
axes[0, 1].set_ylim(0, 10)
axes[0, 1].grid(True, alpha=0.3)

# Bottom-left: Consensus strength (inverse std)
std_per_round = [np.std([SENTIMENT[a][r] for a in AGENTS]) for r in range(5)]
consensus = [1.0 / (1.0 + s) for s in std_per_round]
axes[1, 0].bar(rounds, consensus, color="#2e7d32", alpha=0.8)
axes[1, 0].set_xlabel("Debate Round")
axes[1, 0].set_ylabel("Consensus Strength")
axes[1, 0].set_title("Agent Agreement Over Time (Higher = More Consensus)")
axes[1, 0].grid(True, alpha=0.3)

# Bottom-right: Direction votes per round
up_pct = [sum(1 for a in AGENTS if DIRECTION[a][r] == "UP") / len(AGENTS) for r in range(5)]
flat_pct = [sum(1 for a in AGENTS if DIRECTION[a][r] == "FLAT") / len(AGENTS) for r in range(5)]
down_pct = [sum(1 for a in AGENTS if DIRECTION[a][r] == "DOWN") / len(AGENTS) for r in range(5)]
axes[1, 1].bar(rounds, up_pct, color="#2e7d32", label="UP", alpha=0.8)
axes[1, 1].bar(rounds, flat_pct, bottom=up_pct, color="#ffa000", label="FLAT", alpha=0.8)
axes[1, 1].bar(
    rounds,
    down_pct,
    bottom=[u + f for u, f in zip(up_pct, flat_pct, strict=True)],
    color="#c62828",
    label="DOWN",
    alpha=0.8,
)
axes[1, 1].set_xlabel("Debate Round")
axes[1, 1].set_ylabel("Vote Share")
axes[1, 1].set_title("Direction Votes: Bears Become Bulls")
axes[1, 1].legend()
axes[1, 1].grid(True, alpha=0.3)

plt.suptitle(
    "MineSites SATF Swarm Simulation — 8 Experts x 5 Rounds",
    fontsize=14,
    fontweight="bold",
    y=1.01,
)
plt.tight_layout()
plt.show()

# Print summary stats
print("\nRound 1 → Round 5 sentiment shift:")
for agent in AGENTS:
    delta = SENTIMENT[agent][-1] - SENTIMENT[agent][0]
    print(f"  {agent:25s}: {SENTIMENT[agent][0]} → {SENTIMENT[agent][-1]} ({delta:+d})")
print(
    f"\n  Swarm average: {avg_sentiment[0]:.1f} → {avg_sentiment[-1]:.1f} ({avg_sentiment[-1] - avg_sentiment[0]:+.1f})"
)
print(f"  Final bullish ratio: {up_pct[-1]:.0%}")
print(f"  Final consensus: {consensus[-1]:.3f}")
```
